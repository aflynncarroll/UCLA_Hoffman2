```bash
#!/bin/bash

# Define the run directory
run_dir="/u/project/geschwind/shared/GenomicDatasets-processed/Imputed/"

# Iterate over all subdirectories in the run directory
for dir in "${run_dir}"*/; do
  # Exclude directories named "total_vcfs_r2_0.3"
  if [ "$(basename "$dir")" != "total_vcfs_r2_0.3" ]; then
    # Extract the subdirectory name without the trailing slash
    sub_dir_name=$(basename "$dir")

    # Define the full path to the output file
    output_file="${run_dir}total_vcfs_r2_0.3/${sub_dir_name}_total.vcf.gz"

    # Concatenate VCF files, filter, normalize, and annotate
    bcftools concat "${dir}"chr*.dose.vcf.gz -Ou |
      bcftools view -Ou -i 'R2>0.3' |
      bcftools norm -Ou -m -any |
      bcftools norm -Ou -f /u/project/pasaniuc/afcarrol/Projects/20230906_imputed_data_qc/hg38.fa |
      bcftools annotate -Oz -o "$output_file"
  fi
done
```

```bash
# Directory containing .vcf.gz files
vcf_dir="/path/to/your/directory/"

# Define the output CSV file
output_csv="sample_names.csv"

# Create a header for the CSV file
echo "File Name,Sample Name" > "$output_csv"

# Loop through each .vcf.gz file in the directory
for vcf_file in "${vcf_dir}"*.vcf.gz; do
  # Check if the file exists
  if [ -e "$vcf_file" ]; then
    # Get the file name without the path
    file_name=$(basename "$vcf_file")
    
    # Extract sample names from the VCF file
    sample_names=$(bcftools query -l "$vcf_file")
    
    # Create a line for each sample in the CSV file
    for sample in $sample_names; do
      echo "$file_name,$sample" >> "$output_csv"
    done
  else
    echo "File not found: $vcf_file"
  fi
done
```
