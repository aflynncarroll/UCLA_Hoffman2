```r
## ggplot2 classic theme defaults' edits:
base_size <- 6 # base text size
theme_set(theme_classic())
theme_update(
  # Base text
  text = element_text(family = "Helvetica", size = base_size), # placeholder: I prefer to change font family to Arial outside R
  # Axes
  axis.title = element_text(face = "bold", size = 8),
  axis.line = element_line(color="black", size = .25),
  axis.text.y = element_text(color="black", size = 6),
  axis.text.x = element_text(color="black", size = 6),
  axis.ticks = element_line(color="black", size = .25),
  # Plot title
  plot.title = element_text(face = "bold", size = 10),
  # Legend(s)
  legend.text = element_text(size = 6),
  legend.key.size = unit(0.4, "lines"),
  legend.box.spacing = unit(0, "cm"),
  # Plot tag
  plot.tag = element_text(face = "bold", size = 10),
  # Plot margin
  plot.margin = margin(0.25, 0.25, 0.25, 0.25, "in"),
  # Facets
  strip.text = element_text(face = "bold", size = 8),
  strip.background = element_rect(size = .25)
)

## ggplot2 specific geom defaults' edits:
update_geom_defaults("hline", list(size = .25))
update_geom_defaults("rect", list(size = .25))
update_geom_defaults("errorbar", list(size = .25))
update_geom_defaults("segment", list(size = .25))
update_geom_defaults("point", list(size = 1))
update_geom_defaults("linerange", list(size = .25))
update_geom_defaults("label_repel", list(segment.size = .25, size = base_size/.pt))
update_geom_defaults("text", list(size = base_size/.pt))
```

```r

all_results <- data.frame()

# Loop over chromosomes 1 to 22
for (chr in 1:22) {
    print(chr)
  # Construct the path to the result file for the current chromosome
  #result_file <- sprintf("/u/home/a/afcarrol/project-pasaniuc/Projects/20240403_spark_gwas_larger_unre/output_files/asd_afr_gwas_chr%d.ASD.glm.logistic", chr)
  result_file <- sprintf("../20240403_spark_gwas_larger_unre/output_files/asd_afr_gwas_chr%d.ASD.glm.logistic", chr)
  # Read the result file
  result <- read.table(result_file, sep = "\t", header = FALSE)

  # Filter results where V7 column is equal to "ADD"
  filtered_results <- result[result$V7 == "ADD", ]

  # Add the filtered results to the data frame for all chromosomes
  all_results <- rbind(all_results, filtered_results)
}

# Print a summary of the combined results
summary(all_results)
```

```r
colnames(all_results) <- c("CHROM", "POS", "ID", "REF", "ALT", "A1", "TEST", "OBS_CT", "OR", "LOG(OR)_SE", "Z_STAT", "P")


all_results <- all_results[complete.cases(all_results$P), ]
```

```r
# Specify the JPEG file path
pdf_file <- "manhattan_plot_spark_afr_asd_4.pdf"

# Open a JPEG device
#jpeg(jpeg_file, width = 31, height = 21, units="cm", res=500)
pdf(pdf_file, width = 6, height = 1.2)
# Create a Manhattan plot within the JPEG device
manhattan(all_results, chr = "CHROM", bp = "POS", snp = "ID", p = "P", 
          col = c("black", "grey"), main = "ASD - SPARK AFR Unrelated")

# Close the JPEG device
dev.off()
```
