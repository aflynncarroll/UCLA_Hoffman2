```r
# Specify the path to the directory
base_path <- "/u/project/geschwind/shared/GenomicDatasets-processed/ACE_X_chroms"

# Get a list of all subdirectories
dir_list <- list.dirs(base_path, full.names = TRUE, recursive = FALSE)
dir_list
```

```r
file.path(dir_list[1], "*.sexcheck")
```

```r

# Initialize an empty list to store data frames
data_list <- list()

sexcheck_files <- sex_files

# Loop through each .sexcheck file found in the current directory
for (file in sexcheck_files) {
    # Read the file
    input <- read.table(file, header = FALSE, fill = TRUE)

    # Set the first row as column names and remove it
    colnames(input) <- as.character(unlist(input[1, ]))
    input <- input[-1, ]  # Remove the first row

    # Add the processed data frame to the list
    data_list[[length(data_list) + 1]] <- input
}

# Combine all data frames into one
combined_data <- do.call(rbind, data_list)
```
