```r
library(tidyverse)
```

```r
# Set the directory path containing the files
directory_path <- "/u/home/a/afcarrol/project-pasaniuc/Projects/20240229_ace_partial_pgs/out/"
# Initialize an empty data frame to store the results
result <- data.frame()
```

```r

# Loop through files from ace_chr1 to ace_chr22
for (chr in 1:22) {
    
#    chr <- 1
  # Construct the file path
  file_path <- paste0(directory_path, "ace_chr", chr, ".sample_pgs.tsv")
  print(file_path)
  # Read the data from the current file
  data <- read.table(file_path, header = TRUE, , sep = "\t")
  
  # Combine the results for each ancestry (ANC1 and ANC2)
  result <- rbind(result, data)
}

```

```r
result[result$indiv == "AU003305_AU003305",]
```

```r
# Sum the scores across all chromosomes for each individual by ancestry
final_result <- result %>%
  group_by(indiv) %>%
  summarize(
    ANC1_sum = sum(ANC1, na.rm = TRUE),
    ANC2_sum = sum(ANC2, na.rm = TRUE)
  )
```
