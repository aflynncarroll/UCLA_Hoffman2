```python
hm3_snps = pyreadr.read_r("map.rds")[None]
```

```python
hm3_snps = hm3_snps[["chr", "pos_hg38"]].rename(columns={"chr": "CHROM", "pos_hg38": "POS"})
```

```python
# create a new column that combines CHROM and POS
hm3_snps['CHROM_POS'] = hm3_snps['CHROM'].astype(str) + ':' + hm3_snps['POS'].astype(str)

```

```python
# write the CHROM_POS column to a text file
hm3_snps['CHROM_POS'].to_csv('hm3_snp_pos.txt', sep='\t', index=False, header=False)
```
