```r

# Split the column into two based on the "." delimiter
split_df <- do.call(rbind, strsplit(as.character(eqtl$gene_id), "\\."))

# Combine the split columns back into the data frame
eqtl$ensgid <- split_df[, 1]
eqtl$ensgid_v <- split_df[, 2]
```

```r
head(eqtl)
```

```r
# Connect to Ensembl
ensembl <- useMart("ensembl", dataset = "hsapiens_gene_ensembl")

# The gene ID (strip version if necessary)
ensembl_ids <- unique(eqtl$ensgid)
# Get HGNC gene symbols
gene_ids <- getBM(attributes = c('ensembl_gene_id', 'hgnc_symbol'),
                filters = 'ensembl_gene_id',
                values = ensembl_ids,
                mart = ensembl)
```
