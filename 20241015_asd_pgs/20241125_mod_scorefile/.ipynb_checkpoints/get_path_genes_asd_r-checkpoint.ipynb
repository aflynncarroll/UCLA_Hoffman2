```r
clinvar <- fread("clinvar_result.txt", sep = "\t")
```

```r
colnames(clinvar)
```

```r
head(clinvar$'dbSNP ID')
```

```r
unique(clinvar$"Germline classification")
```

```r
# Extract unique entries from the "Germline classification" column
unique_entries <- unique(clinvar$`Germline classification`)

# Ensure only entries with "pathogenic" and not "Conflicting classifications of pathogenicity"
pathogenic_entries <- unique_entries[
  grepl("\\bpathogenic\\b", unique_entries, ignore.case = TRUE) & 
  !grepl("Conflicting classifications of pathogenicity", unique_entries, ignore.case = TRUE)
]

pathogenic_entries
```

```r
clin_path <- clinvar[clinvar$`Germline classification` %in% pathogenic_entries,]
```

```r
dim(clin_path)
```

```r
length(unique(clin_path$'Gene(s)'))
```

```r
head(clin_path$'Gene(s)')
```

```r
length(unique(clin_path$'dbSNP ID'))
```

```r
unique(clin_path$'Condition(s)')
```

```r
asd_pheno <- unique(grep("Autism spectrum disorder", clin_path$'Condition(s)', ignore.case = TRUE, value = TRUE))
```

```r
asd_pheno
```

```r
clin_path_asd <- clin_path[clin_path$'Condition(s)' %in% asd_pheno, ]
```

```r
dim(clin_path_asd)
```

```r
asd_path_rsid <- unique(clin_path_asd$'dbSNP ID')
```

```r
length(asd_path_rsid)
```

```r
asd_path_rsid <- asd_path_rsid[asd_path_rsid != '']
```
