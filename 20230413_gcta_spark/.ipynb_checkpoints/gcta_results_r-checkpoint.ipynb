```r
prefix <- "test_grm_spark_1000"

ReadGRMBin=function(prefix, AllN=F, size=4){
  sum_i=function(i){ # sum function
    return(sum(1:i))
  }
  BinFileName=paste(prefix,".grm.bin",sep="") # bin file
  NFileName=paste(prefix,".grm.N.bin",sep="") # n bin file
  IDFileName=paste(prefix,".grm.id",sep="") # id file
  id = read.table(IDFileName) # read in the table with inv and family ids
  n=dim(id)[1] # number of individuals - looking at family column (?)
  BinFile=file(BinFileName, "rb"); # set the bin file
  grm=readBin(BinFile, n=n*(n+1)/2, what=numeric(0), size=size) # read in the binary file
  NFile=file(NFileName, "rb"); # set N file
  if(AllN==T){
    N=readBin(NFile, n=n*(n+1)/2, what=numeric(0), size=size)
  }
  else N=readBin(NFile, n=1, what=numeric(0), size=size)
  i=sapply(1:n, sum_i)
  return(list(diag=grm[i], off=grm[-i], id=id, N=N))
    #return(data.frame(diag=grm[i], off=grm[-i], id=id))
}
```

```r
data <- ReadGRMBin("SPARK_GRM/test_grm_spark_1000.part_1000_0002")
```
