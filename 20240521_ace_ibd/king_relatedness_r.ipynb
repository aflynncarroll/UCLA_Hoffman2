```r
king_ace <- read.table("king_ace_related.kin")
```

```r
colnames(king_ace) <- as.character(unlist(king_ace[1, ]))
king_ace <- king_ace[-1, ]
```

```r
king_ace$ID1 <- sub(".*_", "", king_ace$ID1)
king_ace$ID2 <- sub(".*_", "", king_ace$ID2)
```

```r
head(king_ace)
```

```r
king_ace <- king_ace[,-1]
```

```r
head(king_ace)
```

```r
dim(king_ace)
```

```r
sa_covar <- read.csv("ace_spark_total_covar.csv")
```

```r
sa_covar <- sa_covar[sa_covar$sampleset == "ace", c("FID", "IID")]
```

```r
sa_covar$ID1 <- sa_covar$IID
sa_covar$ID2 <- sa_covar$IID
sa_covar$FID1 <-sa_covar$FID
sa_covar$FID2 <-sa_covar$FID
```

```r
ace_rel <- king_ace[king_ace$Kinship > 0, ]
```

```r
dim(ace_rel)
```

```r
ace_2nd <- ace_rel[ace_rel$Kinship >= (1/8),]
```

```r
dim(ace_2nd)
```

```r
ace_2nd <- left_join(ace_2nd, sa_covar[,c("ID1", "FID1")], by = "ID1")
```

```r
ace_2nd <- left_join(ace_2nd, sa_covar[,c("ID2", "FID2")], by = "ID2")
```

```r
head(ace_2nd)
```

```r
ace_2nd <- ace_2nd[ace_2nd$FID1 != ace_2nd$FID2, ]
```

```r
dim(ace_2nd)
```

```r
ace_2nd
```

```r
ace_final_rel <- ace_2nd[ace_2nd$InfType != "Dup/MZ",]
```

```r
ace_final_rel
```
