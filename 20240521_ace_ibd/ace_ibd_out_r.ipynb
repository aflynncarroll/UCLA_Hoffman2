```r
library(tidyverse)
```

```r
ace_ibd <- read.table("ace_ibd.genome")
```

```r
sa_covar <- read.csv("ace_spark_total_covar.csv")
```

```r
head(sa_covar)
```

```r
table(sa_covar$sampleset)
```

```r
sa_covar <- sa_covar[sa_covar$sampleset == "ace", c("FID", "IID")]
```

```r
head(ace_ibd)
```

```r
colnames(ace_ibd) <- as.character(unlist(ace_ibd[1, ]))
ace_ibd <- ace_ibd[-1, ]
```

```r
ace_ibd$IID1 <- sub(".*_", "", ace_ibd$IID1)
```

```r
ace_ibd$IID2 <- sub(".*_", "", ace_ibd$IID2)
```

```r
head(ace_ibd)
```

```r
range(ace_ibd$PI_HAT)
```

```r
dim(ace_ibd)
```

```r
dim(ace_ibd[ace_ibd$PI_HAT >0.05,])
```

```r
ave_rel <- ace_ibd[ace_ibd$PI_HAT >0.05,]
```

```r
length(unique(ave_rel$IID1))
```

```r
sum(ave_rel$IID1 %in% sa_covar$IID)
```

```r
ave_rel <- ave_rel[ave_rel$IID1 %in% sa_covar$IID & ave_rel$IID2 %in% sa_covar$IID,]
```

```r
dim(ave_rel)
```

```r
head(ave_rel)
```

```r
colnames(ave_rel)
```

```r
ave_rel <- ave_rel[,c("IID1", "IID2", "RT", "EZ", "Z0", "Z1", "Z2", "PI_HAT", "PHE", "DST", "PPC", "RATIO")]
```

```r
head(sa_covar)
```

```r
sa_covar$IID1 <- sa_covar$IID
sa_covar$IID2 <- sa_covar$IID
sa_covar$FID1 <-sa_covar$FID
sa_covar$FID2 <-sa_covar$FID
```

```r
head(sa_covar)
```

```r
ave_rel <- left_join(ave_rel, sa_covar[,c("IID1", "FID1")], by= "IID1")
```

```r
ave_rel <- left_join(ave_rel, sa_covar[,c("IID2", "FID2")], by= "IID2")
```
