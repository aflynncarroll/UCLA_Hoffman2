```r
all_results <- data.frame()

# Loop over chromosomes 1 to 22
for (chr in 1:22) {
    print(chr)
  # Construct the path to the result file for the current chromosome
  result_file <- sprintf("/u/home/a/afcarrol/project-pasaniuc/Projects/20231219_SPARK_AFR_ASD_GWAS/output_files/asd_afr_gwas_chr%d.ASD.glm.logistic", chr)

  # Read the result file
  result <- read.table(result_file, sep = "\t", header = FALSE)

  # Filter results where V7 column is equal to "ADD"
  filtered_results <- result[result$V7 == "ADD", ]

  # Add the filtered results to the data frame for all chromosomes
  all_results <- rbind(all_results, filtered_results)
}

# Print a summary of the combined results
summary(all_results)
```

```r
all_results <- read.csv("/u/home/a/afcarrol/project-pasaniuc/Projects/20231219_SPARK_AFR_ASD_GWAS/output_files/ADD_1-22_spark_afr_asd.csv")
```

```r
man_plot <- manhattan(all_results, 
          chr = "CHROM", 
          bp = "POS", 
          snp = "ID", 
          p = "P", 
          col = c("black", "grey"), 
          main = "SPARK AFR ASD Unrelated")
```

```r
pdf_file <- "manhattan_plot_spark_afr_asd_unre.pdf"

pdf(pdf_file, width = 8, height = 6)

# Create a Manhattan plot within the PDF device
manhattan(all_results, 
          chr = "CHROM", 
          bp = "POS", 
          snp = "ID", 
          p = "P", 
          col = c("black", "grey"), 
          main = "SPARK AFR ASD Unrelated")

# Close the PDF device
dev.off()
```

```r
# Specify the JPEG file path
jpeg_file <- "manhattan_plot_spark_afr_asd.jpg"

# Open a JPEG device
jpeg(jpeg_file, width = 31, height = 21, units="cm", res=500)

# Create a Manhattan plot within the JPEG device
manhattan(all_results, chr = "CHROM", bp = "POS", snp = "ID", p = "P", 
          col = c("black", "grey"), main = "ASD - SPARK AFR Unrelated")

# Close the JPEG device
dev.off()
```
