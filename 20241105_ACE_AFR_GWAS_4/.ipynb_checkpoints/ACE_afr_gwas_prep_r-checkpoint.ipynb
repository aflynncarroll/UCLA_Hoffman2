```r
library(tidyverse)
```

```r
ace_pop <- read.csv("20241104_ACE_phenotypes.csv")
```

```r
head(ace_pop)
```

```r
unre_g3 <- read.table("../20240314_ACE_ASD_gwas_larger_pop//ace_afr_unrelated_3.txt")
```

```r
head(unre_g3)
```

```r
dim(unre_g3)
```

```r
unre_g4 <- unre_g3[unre_g3$V1 %in% ace_pop$sp_id,]
```

```r
head(unre_g3$V1)
```

```r
length(unre_g4)
```

```r
ace_afr_run <- ace_pop[ace_pop$sp_id %in% unre_g4, ]
```

```r
dim(ace_afr_run)
```

```r
ace_afr_run <- ace_afr_run[, c("FID", "sp_id", "ASD", "Age", "Sex", "PC1", "PC2", "PC3", "PC4", "PC5", "PC6", "PC7", "PC8", "PC9", "PC10")]
```

```r
colnames(ace_afr_run) <- c("FID", "IID", "ASD", "Age", "Sex", "PC1", "PC2", "PC3", "PC4", "PC5", "PC6", "PC7", "PC8", "PC9", "PC10")
```

```r
ace_afr_run$FID <- 0
```

```r
table(ace_afr_run$ASD)
```

```r
head(ace_afr_run)
```

```r
ace_afr_run$ASD <- ace_afr_run$ASD + 1
```

```r
head(ace_afr_run)
```

```r
# Specify the path for the output file
output_file_path_ace <- "ace_afr_unrelated_4.txt"


unre_afr_ace_iids <- ace_afr_run$IID
# Write the IIDs to a text file
writeLines(unre_afr_ace_iids, con = output_file_path_ace)
```

```r
write.table(ace_afr_run[,c("FID", "IID","PC1", "PC2", "PC3", "PC4", "PC5", "PC6", "PC7", "PC8", "PC9", "PC10")], 
            file = "ace_covar_unre_4.txt", 
            sep = "\t", quote = FALSE, row.names = FALSE, col.names = TRUE)
```

```r
write.table(ace_afr_run[,c("FID", "IID", "ASD")], 
            file = "ace_asd_unre_4.txt", 
            sep = "\t", quote = FALSE, row.names = FALSE, col.names = TRUE)
```
